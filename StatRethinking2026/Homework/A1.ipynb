Can you figure out the number of honest participants that maximizes the number of wats to realize the observed sample (8 out of 10)?

```python
def lost(honest):
    return 8-(honest-2)+2
def won(honest):
    return 10-lost(honest)
def ways(honest):
    return won(honest)**8+lost(honest)**2

for i in range(9):
    print(f"for {i+2} honest participants, there are {ways(i+2)} ways to oberseve 8 out of 10 winning participants")
```
